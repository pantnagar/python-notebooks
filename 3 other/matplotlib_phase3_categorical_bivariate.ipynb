# Matplotlib Phase 3: Categorical & Bivariate Analysis
### Credit Card Risk Analysis Project (Risk Segmentation)

Risk analysis lives and dies on group comparisons: renters vs. homeowners, paid vs.
defaulted, income vs. credit limit. This phase is about comparing *categories* against
each other and *two numeric variables* against each other.

This notebook covers 3 topics:
7. **Bar Charts** — comparing a metric across categorical groups
8. **Stacked Bar Charts** — showing proportions/composition within a category
9. **Scatter Plots** — finding relationships between two numeric variables

**Format:** Each question has a `YOUR CODE HERE` cell to attempt first, followed by a
`Solution` cell. Try your own answer before peeking!

Run the setup cell below first — it builds a synthetic applicant pool with a housing
status, a default outcome, income, credit limit, and credit score.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

n = 600

housing_status = np.random.choice(['Rent', 'Own', 'Mortgage'], size=n, p=[0.35, 0.25, 0.40])

# Default probability varies by housing status (renters riskier, owners safest)
default_prob_map = {'Rent': 0.18, 'Own': 0.05, 'Mortgage': 0.10}
default_prob = np.array([default_prob_map[h] for h in housing_status])
default = np.random.binomial(1, default_prob)

annual_income = np.random.lognormal(mean=10.8, sigma=0.4, size=n)

# Credit limit is loosely driven by income, plus noise
credit_limit = 3000 + annual_income * 0.15 + np.random.normal(0, 2000, size=n)
credit_limit = np.clip(credit_limit, 500, None)

credit_score = np.clip(np.random.normal(680, 50, size=n), 300, 850)

df = pd.DataFrame({
    'Housing_Status': housing_status,
    'Default': default,               # 1 = defaulted, 0 = paid
    'Annual_Income': annual_income,
    'Credit_Limit': credit_limit,
    'Credit_Score': credit_score
})

print(df.shape)
df.head()
```

**Q9 (Capstone).** What matters most to a risk committee usually isn't raw default *counts* (which are biased by how many applicants are in each group) but the default *rate*. Compute `default_rate = (default_counts / total_counts * 100)`, then plot a fully polished bar chart: bar labels showing the percentage (`ax.bar_label(bars, fmt='%.1f%%')`), a bold title, axis labels, and `ax.grid(axis='y', alpha=0.3)`.

```python
# YOUR CODE HERE
```

> importantly for risk work — a **rate** is usually more meaningful than a raw **count**.

**Q13.** Raw counts don't compare cleanly when group sizes differ. Convert `crosstab` to row-wise percentages with `crosstab_pct = crosstab.div(crosstab.sum(axis=1), axis=0) * 100`, then plot the same stacked bar using `crosstab_pct` instead — now every bar reaches exactly 100%, making composition directly comparable across groups.

```python
# YOUR CODE HERE
```

**Solution 13**

```python
crosstab_pct = crosstab.div(crosstab.sum(axis=1), axis=0) * 100

fig, ax = plt.subplots()
ax.bar(crosstab_pct.index, crosstab_pct[0], label='Paid', color='seagreen')
ax.bar(crosstab_pct.index, crosstab_pct[1], bottom=crosstab_pct[0], label='Defaulted', color='tomato')
ax.legend()
plt.show()
```

**Q14.** Repeat Q13, and add a title `"Loan Outcome Composition by Housing Status"`, a y-label `"Percentage of Applicants"`, and place the legend outside the plot area using `ax.legend(loc='upper left', bbox_to_anchor=(1.0, 1.0))`.

```python
# YOUR CODE HERE
```

**Solution 14**

```python
fig, ax = plt.subplots()
ax.bar(crosstab_pct.index, crosstab_pct[0], label='Paid', color='seagreen')
ax.bar(crosstab_pct.index, crosstab_pct[1], bottom=crosstab_pct[0], label='Defaulted', color='tomato')
ax.set_title("Loan Outcome Composition by Housing Status")
ax.set_ylabel("Percentage of Applicants")
ax.legend(loc='upper left', bbox_to_anchor=(1.0, 1.0))
fig.tight_layout()
plt.show()
```

**Q15.** Repeat Q13's percentage stacked bar, then loop over each housing status and use `ax.text()` to write the paid percentage in the middle of the green segment and the defaulted percentage in the middle of the red segment (e.g. vertical position `crosstab_pct[0][h] / 2` for the paid label).

```python
# YOUR CODE HERE
```

**Solution 15**

```python
fig, ax = plt.subplots()
ax.bar(crosstab_pct.index, crosstab_pct[0], label='Paid', color='seagreen')
ax.bar(crosstab_pct.index, crosstab_pct[1], bottom=crosstab_pct[0], label='Defaulted', color='tomato')

for h in crosstab_pct.index:
    paid_pct = crosstab_pct.loc[h, 0]
    default_pct = crosstab_pct.loc[h, 1]
    ax.text(h, paid_pct / 2, f"{paid_pct:.0f}%", ha='center', va='center', color='white')
    ax.text(h, paid_pct + default_pct / 2, f"{default_pct:.0f}%", ha='center', va='center', color='white')

ax.legend()
plt.show()
```

**Q16 (Capstone).** Build the final version: percentage stacked bar, green/red coloring, in-segment percentage labels, bold title, y-label, gridlines on the y-axis only (`ax.grid(axis='y', alpha=0.3)`), and a legend placed outside the plot.

```python
# YOUR CODE HERE
```

**Solution 16**

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.bar(crosstab_pct.index, crosstab_pct[0], label='Paid', color='seagreen')
ax.bar(crosstab_pct.index, crosstab_pct[1], bottom=crosstab_pct[0], label='Defaulted', color='tomato')

for h in crosstab_pct.index:
    paid_pct = crosstab_pct.loc[h, 0]
    default_pct = crosstab_pct.loc[h, 1]
    ax.text(h, paid_pct / 2, f"{paid_pct:.0f}%", ha='center', va='center', color='white')
    ax.text(h, paid_pct + default_pct / 2, f"{default_pct:.0f}%", ha='center', va='center', color='white')

ax.set_title("Loan Outcome Composition by Housing Status", fontsize=13, fontweight='bold')
ax.set_ylabel("Percentage of Applicants")
ax.grid(axis='y', alpha=0.3)
ax.legend(loc='upper left', bbox_to_anchor=(1.0, 1.0))
fig.tight_layout()
plt.show()
```

> **Checkpoint — Section 8:** Stacked bars are built by plotting one segment normally,
> then plotting the next with `bottom=` set to the previous segment's values. Converting
> raw counts to row-wise percentages (`.div(..., axis=0)`) is what makes composition
> directly comparable across groups of very different sizes.

**Q19.** Color each point by whether the applicant defaulted: build a color list with a list comprehension (`'tomato'` if `Default == 1` else `'seagreen'`), pass it to `c=` in `ax.scatter()`, and manually build a legend using two `plt.Line2D` proxy handles (since a single `scatter()` call with a color list doesn't auto-generate one).

**Solution 19**

```python
from matplotlib.lines import Line2D

colors = ['tomato' if d == 1 else 'seagreen' for d in df['Default']]

fig, ax = plt.subplots()
ax.scatter(df['Annual_Income'], df['Credit_Limit'], c=colors)

legend_handles = [
    Line2D([0], [0], marker='o', color='w', markerfacecolor='seagreen', markersize=8, label='Paid'),
    Line2D([0], [0], marker='o', color='w', markerfacecolor='tomato', markersize=8, label='Defaulted')
]
ax.legend(handles=legend_handles)
plt.show()
```

**Q21.** Fit a straight line to `Annual_Income` vs. `Credit_Limit` using `np.polyfit(df['Annual_Income'], df['Credit_Limit'], 1)` to get a slope and intercept, then plot the original scatter plot with that trend line drawn on top in `'black'`.

```python
# YOUR CODE HERE
```

**Solution 21**

```python
slope, intercept = np.polyfit(df['Annual_Income'], df['Credit_Limit'], 1)
x_line = np.linspace(df['Annual_Income'].min(), df['Annual_Income'].max(), 100)
y_line = slope * x_line + intercept

fig, ax = plt.subplots()
ax.scatter(df['Annual_Income'], df['Credit_Limit'], alpha=0.4)
ax.plot(x_line, y_line, color='black', linewidth=2, label='Trend')
ax.legend()
plt.show()

print(f"Slope: {slope:.4f}, Intercept: {intercept:.1f}")
```
