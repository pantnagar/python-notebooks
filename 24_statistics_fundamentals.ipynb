## Part B: Measures of Spread — Range, Variance, Standard Deviation (Q8-Q15)

**Q9** Calculate VARIANCE by hand, step by step: (1) find the mean, (2) find each value's squared deviation from the mean, (3) average those squared deviations. Use `statistics.variance()` (sample variance) to confirm — note it divides by `n-1`, not `n`, which you'll need to match manually.

````python
# YOUR CODE HERE
````

**Solution 9**

````python
import statistics
scores = [720, 610, 800, 590, 700]

mean_val = sum(scores) / len(scores)
squared_devs = [(x - mean_val) ** 2 for x in scores]
variance_manual = sum(squared_devs) / (len(scores) - 1)

variance_builtin = statistics.variance(scores)
print(round(variance_manual, 2), round(variance_builtin, 2))
````

**Q10** Calculate STANDARD DEVIATION by hand: it's just `sqrt(variance)`. Use `math.sqrt()` on your Q9 variance, and confirm against `statistics.stdev(scores)`.

````python
# YOUR CODE HERE
````

**Solution 10**

````python
import statistics
import math
scores = [720, 610, 800, 590, 700]

variance_val = statistics.variance(scores)
stdev_manual = math.sqrt(variance_val)
stdev_builtin = statistics.stdev(scores)
print(round(stdev_manual, 2), round(stdev_builtin, 2))
````

**Q11** Explain in a comment WHY standard deviation is usually reported instead of variance, even though variance is calculated first — think about UNITS. If `scores` are in points, what units is variance in, versus standard deviation?

````python
# YOUR CODE HERE
````

**Solution 11**

````python
# Variance is in SQUARED units (e.g. "points squared"), which has no
# intuitive real-world meaning. Standard deviation takes the square root,
# bringing it back to the ORIGINAL units (points), so you can directly
# compare it to the data itself — e.g. "scores typically vary by about 85
# points from the mean" is meaningful; "the variance is 7225 points-squared"
# is not something anyone intuitively grasps.
print("See comment above")
````

**Q13** Given `scores = [720, 610, 800, 590, 700]` and its mean/stdev from before, calculate a Z-SCORE for the value `800`: `(800 - mean) / stdev`. This tells you how many standard deviations above/below average that value is. Print it.

````python
# YOUR CODE HERE
````

**Solution 13**

````python
import statistics
scores = [720, 610, 800, 590, 700]
mean_val = statistics.mean(scores)
stdev_val = statistics.stdev(scores)

z_score = (800 - mean_val) / stdev_val
print(round(z_score, 2))
````

**Q14** Write a function `z_score(value, data)` that computes the z-score of any value against a dataset (reusing the Q13 formula, computing mean/stdev internally). Apply it to EVERY value in `scores = [720, 610, 800, 590, 700]` using a list comprehension, printing each score alongside its z-score.

````python
# YOUR CODE HERE
````

**Solution 14**

````python
import statistics

def z_score(value, data):
    mean_val = statistics.mean(data)
    stdev_val = statistics.stdev(data)
    return (value - mean_val) / stdev_val

scores = [720, 610, 800, 590, 700]
results = [(s, round(z_score(s, scores), 2)) for s in scores]
for s, z in results:
    print(s, z)
````

**Q15** Using `z_score()` from Q14, flag any value in `scores` with an absolute z-score greater than 1 (i.e. more than 1 standard deviation from the mean, in EITHER direction) as an "outlier" — print which scores qualify.

````python
# YOUR CODE HERE
````

**Solution 15**

````python
import statistics

def z_score(value, data):
    mean_val = statistics.mean(data)
    stdev_val = statistics.stdev(data)
    return (value - mean_val) / stdev_val

scores = [720, 610, 800, 590, 700]
outliers = [s for s in scores if abs(z_score(s, scores)) > 1]
print(outliers)
````

**Q19** A common OUTLIER DETECTION rule: any value below `Q1 - 1.5*IQR` or above `Q3 + 1.5*IQR` is considered an outlier. Using Q18's Q1/Q3/IQR, calculate these two boundary values, then filter `balances` to find which values (if any) qualify as outliers by this rule.

````python
# YOUR CODE HERE
````

**Solution 19**

````python
import statistics
balances = sorted([500, 800, 1200, 1500, 2000, 3000, 5000, 8000, 12000, 20000])
q1, q2, q3 = statistics.quantiles(balances, n=4)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = [b for b in balances if b < lower_bound or b > upper_bound]
print(f"Bounds: [{lower_bound}, {upper_bound}]")
print("Outliers:", outliers)
````

**Q20** Write a function `percentile_rank(value, data)` that returns what PERCENTAGE of the data a given value is greater than or equal to: `(count of values <= value) / total * 100`. Test with `value=3000` against the `balances` list.

````python
balances = sorted([500, 800, 1200, 1500, 2000, 3000, 5000, 8000, 12000, 20000])

def percentile_rank(value, data):
    count_below_or_equal = len([x for x in data if x <= value])
    return count_below_or_equal / len(data) * 100

print(percentile_rank(3000, balances))
````

## Part D: Correlation & Relationships Between Variables (Q21-Q25)

**Q21** Given two lists representing the SAME customers' scores and incomes:
```python
scores = [720, 610, 800, 590, 700]
incomes = [55000, 32000, 68000, 28000, 51000]
```
Calculate the correlation COEFFICIENT by hand using the formula:
`covariance(x,y) / (stdev(x) * stdev(y))`, where covariance = `sum((x_i - mean_x) * (y_i - mean_y)) / (n - 1)`.
Build this step by step and print the final correlation value.

````python
# YOUR CODE HERE
````

**Solution 21**

````python
import statistics

scores = [720, 610, 800, 590, 700]
incomes = [55000, 32000, 68000, 28000, 51000]

mean_scores = statistics.mean(scores)
mean_incomes = statistics.mean(incomes)

products = [(s - mean_scores) * (i - mean_incomes) for s, i in zip(scores, incomes)]
covariance = sum(products) / (len(scores) - 1)

correlation = covariance / (statistics.stdev(scores) * statistics.stdev(incomes))
print(round(correlation, 3))
````

**Q24** Critical caveat: "correlation is not causation." Given two UNRELATED lists that happen to correlate by coincidence, `ice_cream_sales = [10, 20, 30, 40, 50]` and `pool_visits = [12, 22, 28, 41, 53]`, compute their correlation — it will be high. Explain in a comment why a high correlation here does NOT mean ice cream sales CAUSE pool visits.

````python
# YOUR CODE HERE
````

**Solution 24**

````python
import statistics
ice_cream_sales = [10, 20, 30, 40, 50]
pool_visits = [12, 22, 28, 41, 53]
print(round(statistics.correlation(ice_cream_sales, pool_visits), 3))
# Both variables are actually driven by a THIRD factor — hot weather —
# that increases both independently. They move together because they share
# a common cause, not because one causes the other. This is the classic
# "confounding variable" trap: always ask whether a correlated relationship
# might really be explained by something else entirely before concluding
# one thing drives another.
````

**Q28** The big one: given paired data for 10 loan applicants — `scores`, `incomes`, and `defaults` (0/1) — compute a FULL risk analysis:
```python
scores = [720, 610, 800, 590, 700, 650, 780, 600, 690, 810]
incomes = [55000, 32000, 68000, 28000, 51000, 40000, 62000, 30000, 45000, 70000]
defaults = [0, 1, 0, 1, 0, 1, 0, 1, 0, 0]
```
Print: (a) `describe()` of scores, (b) correlation between score and default, (c) correlation between income and default, (d) which correlation is STRONGER (larger absolute value) — is score or income a better single predictor of default in this sample?

````python
# YOUR CODE HERE
````

**Solution 28**

````python
import statistics

def describe(data):
    return {
        "mean": round(statistics.mean(data), 2),
        "median": statistics.median(data),
        "stdev": round(statistics.stdev(data), 2),
    }

scores = [720, 610, 800, 590, 700, 650, 780, 600, 690, 810]
incomes = [55000, 32000, 68000, 28000, 51000, 40000, 62000, 30000, 45000, 70000]
defaults = [0, 1, 0, 1, 0, 1, 0, 1, 0, 0]

print("Score summary:", describe(scores))

score_default_corr = statistics.correlation(scores, defaults)
income_default_corr = statistics.correlation(incomes, defaults)

print("Score vs default correlation:", round(score_default_corr, 3))
print("Income vs default correlation:", round(income_default_corr, 3))

stronger = "Score" if abs(score_default_corr) > abs(income_default_corr) else "Income"
print(f"{stronger} is the stronger single predictor of default in this sample")
````
