**Q15** Given this month's and last month's defaulted customer IDs:
```python
this_month = {101, 102, 105, 107}
last_month = {102, 103, 105}
```
Find: (a) repeat defaulters (both months), (b) new defaulters this month, (c) customers who recovered (defaulted last month but not this month). Print all three clearly labeled.

````python
# YOUR CODE HERE
#a
this_month = {101, 102, 105, 107}
last_month = {102, 103, 105}
this_month & last_month

this_month-last_month

last_month-this_month
````

**Solution 15**

````python
this_month = {101, 102, 105, 107}
last_month = {102, 103, 105}

repeat_defaulters = this_month & last_month
new_defaulters = this_month - last_month
recovered = last_month - this_month

print("Repeat defaulters:", repeat_defaulters)
print("New defaulters:", new_defaulters)
print("Recovered:", recovered)
````

**Q16** Given two datasets about to be merged:
```python
dataset_a_ids = {1, 2, 3, 4, 5}
dataset_b_ids = {4, 5, 6, 7}
```
Before merging, check: will an INNER join keep more or fewer rows than a LEFT join on dataset_a? Print the count of matching IDs (inner join size) and the count of dataset_a-only IDs (rows a left join would keep with nulls).

````python
# YOUR CODE HERE
dataset_a_ids = {1, 2, 3, 4, 5}
dataset_b_ids = {4, 5, 6, 7}
len(dataset_a_ids & dataset_b_ids)
len(dataset_a_ids - dataset_b_ids)
````

**Solution 16**

````python
dataset_a_ids = {1, 2, 3, 4, 5}
dataset_b_ids = {4, 5, 6, 7}

inner_match_count = len(dataset_a_ids & dataset_b_ids)
left_only_count = len(dataset_a_ids - dataset_b_ids)

print("Inner join would keep:", inner_match_count, "matching rows")
print("Left-only (no match in b):", left_only_count, "rows")
````

**Q19** Given a list of transactions with duplicate transaction IDs due to a system glitch: `txn_ids = ["T1", "T2", "T1", "T3", "T2", "T4"]`, write code that prints ONLY the IDs that appear MORE THAN ONCE (not just unique values — the actual duplicates).

````python
# YOUR CODE HERE
txn_ids = ["T1", "T2", "T1", "T3", "T2", "T4"]
````

**Solution 19**

````python
txn_ids = ["T1", "T2", "T1", "T3", "T2", "T4"]

seen = set()
duplicates = set()

for tid in txn_ids:
    if tid in seen:
        duplicates.add(tid)
    seen.add(tid)

print(duplicates)
````
