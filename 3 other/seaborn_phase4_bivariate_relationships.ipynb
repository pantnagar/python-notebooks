# Seaborn Phase 4: Bivariate & Multivariate Relationships
### Credit Card Risk Analysis Project (The Core of Risk Analysis)

Predicting default is fundamentally about seeing how financial behaviors interact. The
single most important tool here is `hue=` — it lets you color-code *any* chart by a
third variable, turning a plain relationship into a group comparison for free.

This notebook covers 2 topics:
8. **Scatterplot with Hue** — two numbers plotted against each other, colored by a third
9. **Jointplot** — a scatterplot in the middle, with each variable's own distribution
   shown along the top and right edges

`Solution` cell. Try your own answer before peeking!

Run the setup cell below first.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
sns.set_theme(style="whitegrid")

n = 600

housing_status = np.random.choice(['Rent', 'Own', 'Mortgage'], size=n, p=[0.35, 0.25, 0.40])

annual_income = np.random.lognormal(mean=10.8, sigma=0.4, size=n)
total_debt = annual_income * np.random.uniform(0.05, 0.55, size=n)
debt_to_income = np.clip((total_debt / annual_income) * 100, 0, 90)

credit_score = np.clip(np.random.normal(660, 65, n), 300, 850)
credit_limit = np.clip(3000 + annual_income * 0.15 + np.random.normal(0, 2000, n), 500, None)

raw_risk = (debt_to_income / 100) * 0.6 + ((850 - credit_score) / 550) * 0.6
default_probability = np.clip(raw_risk + np.random.normal(0, 0.08, n), 0.01, 0.95)
default = np.random.binomial(1, default_probability)

df = pd.DataFrame({
    'Housing_Status': housing_status,
    'Annual_Income': annual_income,
    'Total_Debt': total_debt,
    'Debt_to_Income': debt_to_income,
    'Credit_Score': credit_score,
    'Credit_Limit': credit_limit,
    'Default_Probability': default_probability,
    'Default': default
})

print(df.shape)
df.head()
```

**Solution 8**

```python
slope, intercept = np.polyfit(df['Annual_Income'], df['Total_Debt'], 1)
x_line = np.linspace(df['Annual_Income'].min(), df['Annual_Income'].max(), 100)
y_line = slope * x_line + intercept

ax = sns.scatterplot(data=df, x='Annual_Income', y='Total_Debt', hue='Default',
                      palette={0: 'steelblue', 1: 'tomato'}, alpha=0.6)
ax.plot(x_line, y_line, color='black', linewidth=2, label='Trend')
ax.legend()
plt.title("Income vs. Debt with Trend Line")
plt.show()
```

## Section 9: Jointplot

A scatterplot alone doesn't show you each variable's individual distribution. A
jointplot solves that by combining a central scatterplot (or density plot) with a
histogram of each variable along the top and right edges — the relationship and both
marginal distributions, in one figure.
