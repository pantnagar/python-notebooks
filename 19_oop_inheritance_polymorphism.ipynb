# OOP Part 3 — Inheritance & Polymorphism
### 25 Questions

**The core idea:** Inheritance lets a class REUSE another class's code instead of copy-pasting it, while adding or changing specific parts. Polymorphism means different classes can respond to the SAME method call in their own way — you call `.calculate_fee()` on any account type and don't need to know or care which exact type it is.

Think of it like this: `Account` is the general blueprint. `SavingsAccount` and `CreditCardAccount` are BOTH "a kind of" `Account`, but each has its own twist. That "is-a" relationship is the test for whether inheritance is the right tool.

**Solution 23**

```python
class LoanProduct:
    def __init__(self, principal, rate):
        self.principal = principal
        self.rate = rate

    def monthly_payment(self, months):
        return (self.principal * (1 + self.rate)) / months

class PersonalLoan(LoanProduct):
    def monthly_payment(self, months):
        return super().monthly_payment(months) + 25 / months

class MortgageLoan(LoanProduct):
    def monthly_payment(self, months):
        return self.principal * self.rate / months

base = LoanProduct(10000, 0.05)
personal = PersonalLoan(10000, 0.05)
mortgage = MortgageLoan(10000, 0.05)

print(round(base.monthly_payment(12), 2))
print(round(personal.monthly_payment(12), 2))
print(round(mortgage.monthly_payment(12), 2))
```

**Q25** The big one: build a base `RiskModel` class with a method `score(self, applicant)` that raises `NotImplementedError` (a common pattern signaling "subclasses MUST override this"). Build TWO subclasses `SimpleScoreModel(RiskModel)` (returns `applicant["credit_score"]` directly) and `WeightedScoreModel(RiskModel)` (returns `applicant["credit_score"] * 0.7 + applicant["income"] / 1000 * 0.3`). Loop through a list of BOTH model types applied to the SAME applicant, printing each model's name and resulting score — and separately, confirm calling `.score()` on a plain `RiskModel()` instance raises the `NotImplementedError`.

```python
# YOUR CODE HERE
```

**Solution 25**

```python
class RiskModel:
    def score(self, applicant):
        raise NotImplementedError("Subclasses must implement score()")

class SimpleScoreModel(RiskModel):
    def score(self, applicant):
        return applicant["credit_score"]

class WeightedScoreModel(RiskModel):
    def score(self, applicant):
        return applicant["credit_score"] * 0.7 + applicant["income"] / 1000 * 0.3

applicant = {"credit_score": 720, "income": 45000}
models = [SimpleScoreModel(), WeightedScoreModel()]

for m in models:
    print(type(m).__name__, m.score(applicant))

base_model = RiskModel()
try:
    base_model.score(applicant)
except NotImplementedError as e:
    print("Error:", e)
```
