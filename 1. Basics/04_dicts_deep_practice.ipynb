**Q18** Using a dict comprehension, build `risk_map` from `scores` where each value is `"Low"` if score>=750, `"Medium"` if score>=650, else `"High"`.

````python
# YOUR CODE HERE
````

**Solution 18**

## Part E: Mini Challenges (Q26-Q30)

**Q26** Given a list of transaction dicts:
```python
transactions = [
    {"customer": "Alice", "amount": 500},
    {"customer": "Bob", "amount": 1200},
    {"customer": "Alice", "amount": 300},
]
```
Build a dict that sums total transaction amount PER customer (i.e. group and aggregate).

````python
# YOUR CODE HERE
````

**Solution 26**

````python
transactions = [
    {"customer": "Alice", "amount": 500},
    {"customer": "Bob", "amount": 1200},
    {"customer": "Alice", "amount": 300},
]

totals = {}
for t in transactions:
    name = t["customer"]
    totals[name] = totals.get(name, 0) + t["amount"]

print(totals)
````

**Q27** Given `word = "credit risk analysis"`, build a dict counting how many times each character appears (ignore spaces).

````python
# YOUR CODE HERE
````

**Solution 27**

````python
word = "credit risk analysis"
counts = {}
for ch in word:
    if ch == " ":
        continue
    counts[ch] = counts.get(ch, 0) + 1
print(counts)
````

**Q28** Given two dicts representing scores from two different bureaus:
```python
bureau_a = {"Alice": 720, "Bob": 610}
bureau_b = {"Alice": 700, "Bob": 630, "Cara": 800}
```
Build a dict of AVERAGE score per customer. For customers only in one bureau, just use that single score.

````python
# YOUR CODE HERE
````

**Solution 28**

````python
bureau_a = {"Alice": 720, "Bob": 610}
bureau_b = {"Alice": 700, "Bob": 630, "Cara": 800}

all_names = set(bureau_a) | set(bureau_b)
averages = {}
for name in all_names:
    scores = [d[name] for d in (bureau_a, bureau_b) if name in d]
    averages[name] = sum(scores) / len(scores)

print(averages)
````

**Q30** Combine everything: given a list of loan application dicts:
```python
applications = [
    {"name": "Alice", "score": 720, "amount": 15000, "approved": True},
    {"name": "Bob", "score": 610, "amount": 8000, "approved": False},
    {"name": "Cara", "score": 800, "amount": 22000, "approved": True},
    {"name": "Dave", "score": 590, "amount": 5000, "approved": False},
]
```
Calculate: (a) approval rate as a percentage, (b) total approved loan amount, (c) average score of approved applicants only. Print all three.

````python
# YOUR CODE HERE
````

**Solution 30**

````python
applications = [
    {"name": "Alice", "score": 720, "amount": 15000, "approved": True},
    {"name": "Bob", "score": 610, "amount": 8000, "approved": False},
    {"name": "Cara", "score": 800, "amount": 22000, "approved": True},
    {"name": "Dave", "score": 590, "amount": 5000, "approved": False},
]

approved = [a for a in applications if a["approved"]]

approval_rate = len(approved) / len(applications) * 100
total_approved_amount = sum(a["amount"] for a in approved)
avg_approved_score = sum(a["score"] for a in approved) / len(approved)

print(f"Approval rate: {approval_rate:.1f}%")
print(f"Total approved amount: ${total_approved_amount:,}")
print(f"Average score (approved): {avg_approved_score:.1f}")
````
