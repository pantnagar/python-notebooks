## Section 4: Broadcasting

Run this setup cell first — `monthly_balances` is a 3-customer x 3-month matrix:

```python
monthly_balances = np.array([
    [1200, 1300, 1250],
    [400, 420, 410],
    [8900, 9100, 8800],
])
print(monthly_balances.shape)
```

**Q16.** You have one interest rate per month: `interest_rates = np.array([0.02, 0.02, 0.02])`. Use broadcasting to compute `interest_applied` = balance * (1 + rate) for every customer/month, in one line (no loop).

```python
# YOUR CODE HERE
```

**Solution**

```python
interest_rates = np.array([0.02, 0.02, 0.02])
interest_applied = monthly_balances * (1 + interest_rates)
print(interest_applied)
```

**Q17.** Each customer has a different minimum payment: `account_min_payments = np.array([[25], [15], [200]])` (shape (3,1)). Broadcast-subtract this from `monthly_balances` to get `utilization_after_min`, the balance remaining after minimum payment each month.

```python
# YOUR CODE HERE
```

**Solution**

```python
account_min_payments = np.array([[25], [15], [200]])
utilization_after_min = monthly_balances - account_min_payments
print(utilization_after_min)
```

**Q18.** Compute each customer's average balance across the 3 months (`avg_balance_per_account`, shape (3,1), use `keepdims=True`), then broadcast-subtract it from `monthly_balances` to get `deviation_from_avg` — how far each month's balance is from that customer's own average.

```python
# YOUR CODE HERE
```

**Solution**

```python
avg_balance_per_account = monthly_balances.mean(axis=1, keepdims=True)
deviation_from_avg = monthly_balances - avg_balance_per_account
print(avg_balance_per_account)
print(deviation_from_avg)
```

**Q19.** Compute the max balance in each month (`col_max`, across customers — i.e. `axis=0`), then broadcast-divide `monthly_balances` by `col_max` to get `normalized`, where each month's values are scaled 0-1 relative to that month's highest balance.

```python
# YOUR CODE HERE
```

**Solution**

```python
col_max = monthly_balances.max(axis=0)
normalized = monthly_balances / col_max
print(col_max)
print(normalized)
```
