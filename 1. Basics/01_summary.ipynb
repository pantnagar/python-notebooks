**Q2.3** Using a `while` loop, simulate a debt of 5000 being paid off at 500/month.
Print the remaining balance after each payment, and how many months it took.

````python
# YOUR CODE HERE
````

**Solution 2.3**

````python
balance = 5000
months = 0

while balance > 0:
    balance -= 500
    months += 1
    print(f"Month {months}: Remaining balance = {max(balance,0)}")

print(f"Paid off in {months} months")
````

## Topic 3: Functions

**Q3.1** Write a function `classify_risk(score)` that returns "Low Risk", "Medium Risk", or "High Risk"
using the same thresholds as before. Then use it with `map()` on a list of scores.

````python
scores = [810, 640, 700, 590, 760]

# YOUR CODE HERE
````

**Solution 3.1**

````python
def classify_risk(score):
    if score >= 750:
        return "Low Risk"
    elif score >= 650:
        return "Medium Risk"
    else:
        return "High Risk"

scores = [810, 640, 700, 590, 760]
results = list(map(classify_risk, scores))
print(results)
````

**Q3.2** Write a function `monthly_payment(principal, annual_rate, months)` that calculates
a simple monthly payment estimate: `(principal + (principal * annual_rate/100)) / months`.
Give it a default value of `months=12`. Call it two ways: with and without specifying months.

````python
# YOUR CODE HERE
````

**Solution 3.2**

````python
def monthly_payment(principal, annual_rate, months=12):
    total = principal + (principal * annual_rate / 100)
    return total / months

print(monthly_payment(10000, 5))
print(monthly_payment(10000, 5, months=24))
````

## Topic 6: Error Handling

**Q6.1** Write a function `safe_divide_risk(defaults, total_loans)` that returns the default rate
(`defaults/total_loans`) but handles the case where `total_loans` is 0 by returning `None`
instead of crashing. Use try/except.

````python
# YOUR CODE HERE
````

**Solution 6.1**

````python
def safe_divide_risk(defaults, total_loans):
    try:
        return defaults / total_loans
    except ZeroDivisionError:
        return None

print(safe_divide_risk(5, 100))
print(safe_divide_risk(5, 0))
````

## Topic 7: Basic OOP

**Q7.1** Create a `Customer` class with attributes `name`, `score`, `balance`.
Add a method `risk_category()` that returns the risk level (reuse Q2.1 logic).
Create two `Customer` objects and print their risk categories.

````python
# YOUR CODE HERE
````

**Solution 7.1**

````python
class Customer:
    def __init__(self, name, score, balance):
        self.name = name
        self.score = score
        self.balance = balance

    def risk_category(self):
        if self.score >= 750:
            return "Low Risk"
        elif self.score >= 650:
            return "Medium Risk"
        else:
            return "High Risk"

c1 = Customer("Alice", 720, 15000)
c2 = Customer("Bob", 610, 8000)

print(c1.name, c1.risk_category())
print(c2.name, c2.risk_category())
````
