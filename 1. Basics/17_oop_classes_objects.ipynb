**Q9** Add a method `withdraw(self, amount)` that DECREASES `self.balance`, but returns `False` (and does NOT change the balance) if `amount > self.balance`; otherwise decreases it and returns `True`. Test both a successful and a failed withdrawal.

````python
class Customer:
    def __init__(self, name, score, balance=0):
        self.name = name
        self.score = score
        self.balance = balance

    def withdraw(self, amount):
        if amount > self.balance:
            return False
        self.balance -= amount
        return True

c = Customer("Alice", 720, 1000)
print(c.withdraw(300))
print(c.balance)
print(c.withdraw(5000))
print(c.balance)
````

**Q17** Add `__lt__(self, other)` (less-than) comparing by `self.score`. Confirm you can now use `sorted()` directly on a list of `Customer` objects WITHOUT a `key=` lambda, since Python now knows how to compare them.

````python
# YOUR CODE HERE
````

**Solution 17**

````python
class Customer:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def __lt__(self, other):
        return self.score < other.score

    def __repr__(self):
        return f"Customer({self.name}, {self.score})"

customers = [Customer("Alice", 720), Customer("Bob", 610), Customer("Cara", 800)]
print(sorted(customers))
````

**Q23** Build a `LoanAccount` class with `__init__(self, owner, balance, interest_rate)`. Add methods `apply_interest(self)` (multiplies balance by `1 + interest_rate`, modifies in place) and `__str__(self)` for readable printing. Create an account, apply interest twice, printing the object after each application.

````python
# YOUR CODE HERE
````

**Solution 23**

````python
class LoanAccount:
    def __init__(self, owner, balance, interest_rate):
        self.owner = owner
        self.balance = balance
        self.interest_rate = interest_rate

    def apply_interest(self):
        self.balance *= (1 + self.interest_rate)

    def __str__(self):
        return f"{self.owner}: ${self.balance:,.2f}"

acc = LoanAccount("Alice", 10000, 0.05)
acc.apply_interest()
print(acc)
acc.apply_interest()
print(acc)
````

**Q24** Build a `Portfolio` class that holds a list of `Customer` objects (reuse the `Customer` class with name/score from earlier) and add a method `average_score(self)` that computes the average score ACROSS all held customers. Create a portfolio with 3 customers and call `.average_score()`.

````python
# YOUR CODE HERE
````

**Solution 24**

````python
class Customer:
    def __init__(self, name, score):
        self.name = name
        self.score = score

class Portfolio:
    def __init__(self, customers):
        self.customers = customers

    def average_score(self):
        total = sum(c.score for c in self.customers)
        return total / len(self.customers)

portfolio = Portfolio([Customer("Alice", 720), Customer("Bob", 610), Customer("Cara", 800)])
print(portfolio.average_score())
````

**Q25** The big one: build a `CreditApplication` class with `__init__(self, name, score, income, requested_amount)`. Add a method `decide(self)` implementing: score<600 → "Declined"; score>=700 and income>=40000 → "Approved"; else → "Review". Add `__str__` showing name and decision. Create a LIST of 4 applications with varied data, loop through printing each one's string representation.

````python
# YOUR CODE HERE
````

**Solution 25**

````python
class CreditApplication:
    def __init__(self, name, score, income, requested_amount):
        self.name = name
        self.score = score
        self.income = income
        self.requested_amount = requested_amount

    def decide(self):
        if self.score < 600:
            return "Declined"
        if self.score >= 700 and self.income >= 40000:
            return "Approved"
        return "Review"

    def __str__(self):
        return f"{self.name}: {self.decide()}"

applications = [
    CreditApplication("Alice", 720, 45000, 15000),
    CreditApplication("Bob", 580, 30000, 8000),
    CreditApplication("Cara", 680, 35000, 22000),
    CreditApplication("Dave", 810, 60000, 5000),
]

for app in applications:
    print(app)
````
