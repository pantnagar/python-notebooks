## Section 5: Random & Simulation

Modern numpy prefers `np.random.default_rng()` over the legacy `np.random.seed()` API — it's more predictable and each generator instance is independent.

**Q19.** Create a random generator `rng = np.random.default_rng(42)`. Simulate 10 potential loss amounts from a normal distribution with mean 5000 and std 1500, store as `simulated_losses`.

```python
# YOUR CODE HERE
```

**Solution**

```python
rng = np.random.default_rng(42)
simulated_losses = rng.normal(loc=5000, scale=1500, size=10)
print(simulated_losses)
```

## Section 7: Performance & Vectorization

This is *why* numpy matters for real datasets: vectorized operations run in compiled C loops instead of slow Python loops.

**Q25.** Given `big_array = np.arange(1_000_000)`, time how long it takes to compute `sum(val * 2 for val in big_array[:100000])` using a plain Python `for` loop (store the result as `loop_result` and the elapsed time as `loop_time`), versus doing it fully vectorized: `(big_array[:100000] * 2).sum()` (store as `vectorized_result` and `vector_time`). Print both results (they should match) and both times — the vectorized version should be noticeably faster.

```python
big_array = np.arange(1_000_000)

# YOUR CODE HERE
```

**Solution**

```python
big_array = np.arange(1_000_000)

start = time.time()
loop_result = 0
for val in big_array[:100000]:
    loop_result += val * 2
loop_time = time.time() - start

start = time.time()
vectorized_result = (big_array[:100000] * 2).sum()
vector_time = time.time() - start

print(loop_result == vectorized_result, "loop:", loop_time, "vector:", vector_time)
```

**Q26.** Given `monthly_txns` below (3 customers x 4 months), write a function `volatility(row)` that returns `row.max() - row.min()`, then apply it to every row using `np.apply_along_axis` (axis=1) to get `row_volatility` — one volatility value per customer.

```python
monthly_txns = np.array([
    [100, 200, 150, 300],
    [50, 60, 40, 80],
    [900, 850, 920, 880],
])

# YOUR CODE HERE
```

**Solution**

```python
def volatility(row):
    return row.max() - row.min()

row_volatility = np.apply_along_axis(volatility, 1, monthly_txns)
print(row_volatility)
```
