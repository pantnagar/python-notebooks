## Part E: Mini Challenges (Q28-Q30)

**Q28** Given customer records as a list of tuples:
```python
records = [("Alice", 720, 15000), ("Bob", 610, 8000), ("Cara", 800, 22000)]
```
Loop through and print a formatted line for each: name, risk category (>=750 Low, >=650 Medium, else High), and balance formatted as currency.

````python
# YOUR CODE HERE
````

**Solution 28**

````python
records = [("Alice", 720, 15000), ("Bob", 610, 8000), ("Cara", 800, 22000)]

for name, score, balance in records:
    if score >= 750:
        risk = "Low"
    elif score >= 650:
        risk = "Medium"
    else:
        risk = "High"
    print(f"{name}: {risk} Risk, Balance: ${balance:,.2f}")
````

**Q29** Given `records` from Q28, build a list comprehension that returns only the names of customers with "High" risk (score < 650).

````python
# YOUR CODE HERE
````

**Solution 29**

````python
records = [("Alice", 720, 15000), ("Bob", 610, 8000), ("Cara", 800, 22000)]
high_risk_names = [name for name, score, balance in records if score < 650]
print(high_risk_names)
````

**Q30** Given two lists representing this month's and last month's defaulted customer IDs:
```python
this_month = [101, 102, 105, 107]
last_month = [102, 103, 105]
```
Find: (a) customers who defaulted in both months, (b) customers who are NEW defaulters this month (in this_month but not last_month). Use set operations.

````python
# YOUR CODE HERE
````

**Solution 30**

````python
this_month = [101, 102, 105, 107]
last_month = [102, 103, 105]

both_months = set(this_month) & set(last_month)
new_defaulters = set(this_month) - set(last_month)

print("Defaulted both months:", both_months)
print("New defaulters:", new_defaulters)
````
