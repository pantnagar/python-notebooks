## Part D: Raising Exceptions & Custom Exceptions (Q20-Q25)

**Q20** Write a function `validate_score(score)` that RAISES a `ValueError` with the message `"Score must be between 300 and 850"` if score is out of that range, otherwise returns the score. Call it once with a valid score and once with an invalid one wrapped in try/except to catch and print your own raised error.

````python
# YOUR CODE HERE
````

**Solution 20**

````python
def validate_score(score):
    if not (300 <= score <= 850):
        raise ValueError("Score must be between 300 and 850")
    return score

print(validate_score(720))

try:
    validate_score(1000)
except ValueError as e:
    print("Caught:", e)
````

**Q21** Define a CUSTOM exception class `InsufficientFundsError` that inherits from `Exception`. Write a function `withdraw(balance, amount)` that raises it (with a descriptive message including the shortfall) if `amount > balance`. Test it in a try/except.

````python
# YOUR CODE HERE
````

**Solution 21**

````python
class InsufficientFundsError(Exception):
    pass

def withdraw(balance, amount):
    if amount > balance:
        shortfall = amount - balance
        raise InsufficientFundsError(f"Short by {shortfall}")
    return balance - amount

try:
    withdraw(100, 250)
except InsufficientFundsError as e:
    print("Withdrawal failed:", e)
````

**Q22** Define a custom exception `InvalidCreditScoreError(Exception)` that stores the invalid score as an attribute (`self.score = score`) in a custom `__init__`. Raise it from a function, and in the except block, print BOTH the message and access `e.score` directly.

````python
# YOUR CODE HERE
````

**Solution 22**

````python
class InvalidCreditScoreError(Exception):
    def __init__(self, score):
        self.score = score
        super().__init__(f"Invalid credit score: {score}")

def check_score(score):
    if score < 300 or score > 850:
        raise InvalidCreditScoreError(score)
    return "Valid"

try:
    check_score(1200)
except InvalidCreditScoreError as e:
    print(e)
    print("The bad score was:", e.score)
````

**Solution 23**

````python
class LowScoreError(Exception):
    pass

class LowIncomeError(Exception):
    pass

def apply_for_loan(score, income):
    if score < 600:
        raise LowScoreError("Score below minimum threshold")
    if income < 20000:
        raise LowIncomeError("Income below minimum threshold")
    return "Approved"

try:
    result = apply_for_loan(720, 15000)
    print(result)
except LowScoreError as e:
    print("Rejected (score):", e)
except LowIncomeError as e:
    print("Rejected (income):", e)
````

**Solution 24**

````python
class DataParsingError(Exception):
    pass

def parse_score(text):
    try:
        return int(text)
    except ValueError as e:
        raise DataParsingError(f"Could not parse score from '{text}'") from e

try:
    parse_score("bad_data")
except DataParsingError as e:
    print("Final error:", e)
    print("Original cause:", e.__cause__)
````

**Q28** The big one — full validation pipeline. Given raw application rows as strings (some malformed):
```python
raw_rows = [
    "Alice,720,45000",
    "Bob,abc,30000",
    "Cara,800,60000",
    "Dave,650",
    "Eve,700,0",
]
```
Write a function `parse_application(row)` that splits the row, converts score/income to int, and RAISES a custom `MalformedRowError` (with a message including the raw row) if the split doesn't produce exactly 3 parts OR if int conversion fails (catch the underlying error and re-raise as `MalformedRowError` using `from`). Loop through all rows, collecting successfully parsed applications in one list and (row, error_message) pairs for failures in another. Print both lists.

````python
# YOUR CODE HERE
````

**Solution 28**

````python
class MalformedRowError(Exception):
    pass

def parse_application(row):
    parts = row.split(",")
    if len(parts) != 3:
        raise MalformedRowError(f"Expected 3 fields, got {len(parts)}: '{row}'")
    name, score_str, income_str = parts
    try:
        score = int(score_str)
        income = int(income_str)
    except ValueError as e:
        raise MalformedRowError(f"Bad number in row '{row}'") from e
    return {"name": name, "score": score, "income": income}

raw_rows = [
    "Alice,720,45000",
    "Bob,abc,30000",
    "Cara,800,60000",
    "Dave,650",
    "Eve,700,0",
]

valid = []
errors = []

for row in raw_rows:
    try:
        valid.append(parse_application(row))
    except MalformedRowError as e:
        errors.append((row, str(e)))

print("Valid applications:", valid)
print("Errors:", errors)
````
