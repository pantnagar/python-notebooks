````python
# YOUR CODE HERE
````

**Solution 5**

````python
def deep_sum(nested):
    total = 0
    for item in nested:
        if isinstance(item, list):
            total += deep_sum(item)
        else:
            total += item
    return total

data = [1, [2, 3, [4, 5]], 6, [[7]]]
print(deep_sum(data))
````

````python
# YOUR CODE HERE
````

**Solution 11**

````python
def safe_average(numbers):
    try:
        return sum(numbers) / len(numbers)
    except ZeroDivisionError:
        return None

print(safe_average([10, 20, 30]))
print(safe_average([]))
````

Return the FIRST non-None action, checking in order, or `"Allow"` if none trigger. Define 2 policy functions and test one that triggers on the second step.

````python
# YOUR CODE HERE
````

**Solution 17**

````python
def step_block_large_foreign(t):
    if t["amount"] > 3000 and t.get("is_foreign"):
        return "Block"
    return None

def step_review_night(t):
    if t.get("is_night"):
        return "Review"
    return None

def apply_policy(transaction, policy_steps):
    for step in policy_steps:
        action = step(transaction)
        if action is not None:
            return action
    return "Allow"

txn = {"amount": 500, "is_foreign": False, "is_night": True}
print(apply_policy(txn, [step_block_large_foreign, step_review_night]))
````

**Q20** Write `retry_until_confirmed(check_fn, max_attempts=4)` calling zero-argument `check_fn` up to `max_attempts` times, stopping as soon as it returns `True`, returning the ATTEMPT NUMBER it succeeded on, or `-1` if never. Use a closure-based `check_fn` that "succeeds" only on its 3rd call.

````python
# YOUR CODE HERE
````

**Solution 20**

````python
def make_flaky_check():
    calls = [0]
    def check_fn():
        calls[0] += 1
        return calls[0] == 3
    return check_fn

def retry_until_confirmed(check_fn, max_attempts=4):
    for attempt in range(1, max_attempts + 1):
        if check_fn():
            return attempt
    return -1

flaky = make_flaky_check()
print(retry_until_confirmed(flaky))
````

## Full Reference Solution — Mini Project 2
### Don't look until you've attempted all 6 tasks yourself

````python
# Task 1: Score every transaction
for t in transactions:
    t["risk_score"] = round(
        (t["amount"] / 1000) + (2 if t["is_foreign"] else 0) + (1 if t["is_night"] else 0),
        2
    )
print(transactions[1])
````

````python
# Task 2: Flag for review
flagged_ids = [t["txn_id"] for t in transactions if t["risk_score"] > 5]
print(flagged_ids)
````

````python
# Task 3: Per-customer summary
customer_summary = {}
for t in transactions:
    c = t["customer"]
    if c not in customer_summary:
        customer_summary[c] = {"txn_count": 0, "total_amount": 0, "max_risk_score": 0}
    customer_summary[c]["txn_count"] += 1
    customer_summary[c]["total_amount"] += t["amount"]
    customer_summary[c]["max_risk_score"] = max(customer_summary[c]["max_risk_score"], t["risk_score"])

print(customer_summary)
````

````python
# Task 4: Suspicious customers
flagged_customers = {t["customer"] for t in transactions if t["txn_id"] in flagged_ids}

suspicious = sorted([
    c for c, info in customer_summary.items()
    if info["txn_count"] >= 3 and c in flagged_customers
])
print(suspicious)
````

````python
# Task 5: Decision function
def decide_action(txn):
    if txn["risk_score"] > 8:
        return "Block"
    if txn["risk_score"] > 5:
        return "Review"
    if txn["is_foreign"] and txn["amount"] > 1000:
        return "Verify"
    return "Allow"

decisions = {t["txn_id"]: decide_action(t) for t in transactions}
print(decisions)
````

````python
# Task 6: Final report
total_txns = len(transactions)

action_counts = {}
for txn_id, action in decisions.items():
    action_counts[action] = action_counts.get(action, 0) + 1

flagged_amount = sum(t["amount"] for t in transactions if decisions[t["txn_id"]] != "Allow")

top_risk_txn = max(transactions, key=lambda t: t["risk_score"])

report = f"""
FRAUD MONITORING REPORT
========================
Total transactions: {total_txns}
Action breakdown: {action_counts}
Total flagged amount: ${flagged_amount:,.2f}
Highest risk transaction: {top_risk_txn['txn_id']} (score: {top_risk_txn['risk_score']})
"""
print(report)
````
