```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
sns.set_theme(style="whitegrid")

n = 600

housing_status = np.random.choice(['Rent', 'Own', 'Mortgage'], size=n, p=[0.35, 0.25, 0.40])
employment_type = np.random.choice(['Salaried', 'Self-Employed', 'Unemployed'],
                                    size=n, p=[0.6, 0.3, 0.1])

annual_income = np.random.lognormal(mean=10.8, sigma=0.4, size=n)

# Loan amount requested loosely scales with income, with some noise
loan_amount_requested = np.clip(annual_income * np.random.uniform(0.05, 0.30, size=n), 1000, None)
# Renters tend to ask for somewhat smaller loans than homeowners in this dataset
loan_amount_requested = np.where(housing_status == 'Rent',
                                  loan_amount_requested * 0.85, loan_amount_requested)

credit_score = np.clip(np.random.normal(660, 65, n), 300, 850)
debt_to_income = np.clip(np.random.normal(35, 12, n), 0, 90)

raw_risk = (debt_to_income / 100) * 0.6 + ((850 - credit_score) / 550) * 0.6
employment_bump = np.where(employment_type == 'Unemployed', 0.15, 0)
default_probability = np.clip(raw_risk + employment_bump + np.random.normal(0, 0.08, n), 0.01, 0.95)
default = np.random.binomial(1, default_probability)

risk_tier = pd.cut(credit_score, bins=[300, 600, 700, 850], labels=['High', 'Medium', 'Low'])

df = pd.DataFrame({
    'Housing_Status': housing_status,
    'Employment_Type': employment_type,
    'Annual_Income': annual_income,
    'Loan_Amount_Requested': loan_amount_requested,
    'Credit_Score': credit_score,
    'Debt_to_Income': debt_to_income,
    'Default_Probability': default_probability,
    'Default': default,
    'Risk_Tier': risk_tier
})

print(df.shape)
df.head()
```

it computes an estimator (mean by default) per category and automatically adds error
bars showing the uncertainty in that estimate.

**Q12.** Create a 1x2 subplot grid comparing two error bar types for the same mean loan-amount-by-housing chart: left with `errorbar='sd'` (shows one standard deviation of the raw data — how spread out individual applicants are), right with `errorbar=('ci', 95)` (a 95% confidence interval on the *mean* itself — how uncertain the average estimate is). These answer different questions, even though they look similar.

```python
# YOUR CODE HERE
```

**Solution 12**

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5), sharey=True)

sns.barplot(data=df, x='Housing_Status', y='Loan_Amount_Requested', errorbar='sd', ax=axes[0])
axes[0].set_title("Error bars = Standard Deviation")

sns.barplot(data=df, x='Housing_Status', y='Loan_Amount_Requested', errorbar=('ci', 95), ax=axes[1])
axes[1].set_title("Error bars = 95% Confidence Interval")

plt.show()
```
