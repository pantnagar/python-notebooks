## Topic 7: Label-Based Selection (`.loc`)

`.loc` selects by **label** — the values in the index and column names — not position. Run this setup cell first:

```python
df_indexed = df.set_index("application_id")
print(df_indexed.index.name)
df_indexed.head()
```

**Q11.** Using `df_indexed`, get the single row whose label (application ID) is `50001`, into `single_app`, using `.loc[50001]`. Print its `credit_score`.

```python
# YOUR CODE HERE
```

**Solution**

```python
single_app = df_indexed.loc[50001]
print(single_app["credit_score"])
```

**Q12.** `.loc` label-slicing (`.loc[start:end]`) only behaves predictably when the index is sorted — otherwise it slices by whatever order the labels happen to be in, not by ID range. Create `df_sorted = df_indexed.sort_index()`, then get every application with ID between 50001 and 50010 **inclusive** using `.loc[50001:50010]`, into `range_apps`. (Note: unlike `.iloc` slicing, `.loc` slicing includes the endpoint.)

```python
# YOUR CODE HERE
```

**Solution**

```python
df_sorted = df_indexed.sort_index()
range_apps = df_sorted.loc[50001:50010]
print(range_apps.shape)
print(range_apps.index.tolist())
```

**Q13.** From `df_indexed`, select just the `credit_score`, `loan_amount`, and `loan_status` columns for **every** row, using `.loc[:, [...]]`, into `selected_cols_loc`.

```python
# YOUR CODE HERE
```

**Solution**

```python
selected_cols_loc = df_indexed.loc[:, ["credit_score", "loan_amount", "loan_status"]]
print(selected_cols_loc.columns.tolist())
```

**Q14.** This is where `.loc` earns its keep: combine a **boolean condition** for rows with a **column label** in the same call. Get the `loan_status` column, but only for rows where `loan_status` is `"Denied"`, into `denied_status_only`, using `.loc[boolean_condition, "loan_status"]`.

```python
# YOUR CODE HERE
```

**Solution**

```python
denied_status_only = df_indexed.loc[df_indexed["loan_status"] == "Denied", "loan_status"]
print(denied_status_only.shape)
```

**Q15.** `.loc` is also how you **safely modify** a DataFrame in place. Make a copy `df_copy = df_indexed.copy()`, then use `.loc` to set `loan_status` to `"Reviewed"` for every row where `credit_score < 500`. Print how many rows now say `"Reviewed"`.

```python
# YOUR CODE HERE
```

**Solution**

```python
df_copy = df_indexed.copy()
df_copy.loc[df_copy["credit_score"] < 500, "loan_status"] = "Reviewed"
print((df_copy["loan_status"] == "Reviewed").sum())
```

## Topic 9: Multi-Condition Filtering

Real risk rules almost always combine several conditions. **Remember:** use `&` / `|` / `~` (not `and` / `or` / `not`), and wrap each condition in parentheses.

**Q21.** Filter to borrowers who are both low-income (`annual_income < 30000`) **and** taking a large loan (`loan_amount > 10000`), into `risky_low_income_high_loan`. Print the shape.

```python
# YOUR CODE HERE
```

**Solution**

```python
risky_low_income_high_loan = df[(df["annual_income"] < 30000) & (df["loan_amount"] > 10000)]
print(risky_low_income_high_loan.shape)
```

**Q22.** Filter to applications that are **either** `"Denied"` **or** `"Pending"`, into `denied_or_pending`, using `|`. Print the shape.

```python
# YOUR CODE HERE
```

**Solution**

```python
denied_or_pending = df[(df["loan_status"] == "Denied") | (df["loan_status"] == "Pending")]
print(denied_or_pending.shape)
```

**Q23.** Build a more realistic risk rule into `complex_risk`: `credit_score < 600` **AND** (`employment_status == "Unemployed"` **OR** `existing_debt > 15000`). Get the parentheses right — this is exactly the kind of rule where operator precedence bugs hide.

```python
# YOUR CODE HERE
```

**Solution**

```python
complex_risk = df[
    (df["credit_score"] < 600)
    & ((df["employment_status"] == "Unemployed") | (df["existing_debt"] > 15000))
]
print(complex_risk.shape)
```

**Q24.** Filter to every application that is **not** `"Approved"`, into `not_approved`, using `~` to negate the condition.

```python
# YOUR CODE HERE
```

**Solution**

```python
not_approved = df[~(df["loan_status"] == "Approved")]
print(not_approved.shape)
```

**Q25.** Filter to loans between 10000 and 20000 (inclusive) using `.between()` instead of writing two separate `>=`/`<=` conditions, into `mid_range_loans`.

```python
# YOUR CODE HERE
```

**Solution**

```python
mid_range_loans = df[df["loan_amount"].between(10000, 20000)]
print(mid_range_loans.shape)
```

## Topic 10: Membership & Text Filtering

`.isin()` for "is this value one of several options", `.str.contains()` for text search.

**Q26.** Filter to borrowers whose `home_ownership` is `"Own"` **or** `"Mortgage"`, into `owns_or_mortgage`, using `.isin([...])` instead of chained `|` conditions.

```python
# YOUR CODE HERE
```

**Solution**

```python
owns_or_mortgage = df[df["home_ownership"].isin(["Own", "Mortgage"])]
print(owns_or_mortgage.shape)
```

**Q27.** Filter to everyone **except** renters, into `not_renting`, by negating `.isin(["Rent"])` with `~`.

```python
# YOUR CODE HERE
```

**Solution**

```python
not_renting = df[~df["home_ownership"].isin(["Rent"])]
print(not_renting.shape)
```

**Q28.** Filter to every borrower whose `employment_status` contains `"Self"` (case-insensitive), into `self_employed`, using `.str.contains(..., case=False)`.

```python
# YOUR CODE HERE
```

**Solution**

```python
self_employed = df[df["employment_status"].str.contains("Self", case=False)]
print(self_employed.shape)
```

**Q29.** Filter to `employment_status` values that **start with** `"E"`, into `starts_with_e`, using `.str.startswith("E")`. Print the unique values found.

```python
# YOUR CODE HERE
```

**Solution**

```python
starts_with_e = df[df["employment_status"].str.startswith("E")]
print(starts_with_e["employment_status"].unique())
```

**Q30.** `.str.contains()` raises an error on missing (`NaN`) text values by default. Filter to `employment_status` containing `"retired"` (case-insensitive) into `safe_text_filter`, passing `na=False` so any missing values are treated as "no match" instead of crashing. (This column has no actual NaNs, but `na=False` is a habit worth building now — you'll hit real missing text data soon.)

```python
# YOUR CODE HERE
```

**Solution**

```python
safe_text_filter = df[df["employment_status"].str.contains("retired", case=False, na=False)]
print(safe_text_filter.shape)
```
