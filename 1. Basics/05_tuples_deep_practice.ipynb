## Part C: Sorting & Mini Challenges (Q15-Q20)

**Q15** Given `records = [("Alice", 720), ("Bob", 610), ("Cara", 800)]`, sort by score (the second tuple element) ascending using `sorted()` with `key=lambda x: x[1]`.

````python
# YOUR CODE HERE
````

**Solution 15**

````python
records = [("Alice", 720), ("Bob", 610), ("Cara", 800)]
sorted_records = sorted(records, key=lambda x: x[1])
print(sorted_records)
````

**Q16** Sort the same `records` by score DESCENDING (highest score first).

````python
# YOUR CODE HERE
````

**Solution 16**

````python
records = [("Alice", 720), ("Bob", 610), ("Cara", 800)]
sorted_records = sorted(records, key=lambda x: x[1], reverse=True)
print(sorted_records)
````

**Q17** Given `records = [("Alice", 720, 15000), ("Bob", 610, 8000), ("Cara", 800, 22000)]`, sort by score but if two scores tie, sort by balance descending as a secondary key. (Use a tuple key: `key=lambda x: (x[1], -x[2])`)

````python
# YOUR CODE HERE
````

**Solution 17**

````python
records = [("Alice", 720, 15000), ("Bob", 610, 8000), ("Cara", 800, 22000)]
sorted_records = sorted(records, key=lambda x: (x[1], -x[2]), reverse=True)
print(sorted_records)
````

**Q18** Write a function `loan_summary(name, score, amount)` that returns a NAMED tuple-like result using Python's `collections.namedtuple`. Create one instance and access its fields by name (e.g. `result.score`).

````python
# YOUR CODE HERE
````

**Solution 18**

````python
from collections import namedtuple

LoanSummary = namedtuple("LoanSummary", ["name", "score", "amount"])

def loan_summary(name, score, amount):
    return LoanSummary(name, score, amount)

result = loan_summary("Alice", 720, 15000)
print(result.name, result.score, result.amount)
````

**Q19** Given monthly balance snapshots as tuples `(month, balance)`:
```python
history = [("Jan", 5000), ("Feb", 4500), ("Mar", 4000), ("Apr", 4200)]
```
Calculate the month-over-month change for each pair of consecutive months, printing e.g. `"Jan->Feb: -500"`.

````python
# YOUR CODE HERE
````

**Solution 19**

````python
history = [("Jan", 5000), ("Feb", 4500), ("Mar", 4000), ("Apr", 4200)]

for i in range(len(history) - 1):
    month1, bal1 = history[i]
    month2, bal2 = history[i + 1]
    change = bal2 - bal1
    print(f"{month1}->{month2}: {change:+}")
````

**Q20** Combine it all: given a list of applicant tuples `(name, score, requested_amount)`:
```python
applicants = [
    ("Alice", 720, 15000),
    ("Bob", 610, 8000),
    ("Cara", 800, 22000),
    ("Dave", 590, 5000),
]
```
Build a list of tuples `(name, decision)` where decision is `"Approved"` if score >= 650 AND requested_amount <= 20000, else `"Declined"`. Print the result sorted by name.

````python
# YOUR CODE HERE
````

**Solution 20**

````python
applicants = [
    ("Alice", 720, 15000),
    ("Bob", 610, 8000),
    ("Cara", 800, 22000),
    ("Dave", 590, 5000),
]

decisions = [
    (name, "Approved" if score >= 650 and amount <= 20000 else "Declined")
    for name, score, amount in applicants
]

decisions_sorted = sorted(decisions, key=lambda x: x[0])
print(decisions_sorted)
````
