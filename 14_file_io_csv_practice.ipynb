````python
# YOUR CODE HERE
````

**Solution 10**

````python
# with open(...) as f: guarantees the file is closed even if an exception
# is raised inside the block — Python calls f.close() automatically as part
# of the context manager protocol (__exit__), regardless of how the block exits.
#
# With manual open()/close(), if an error happens between the two calls,
# .close() is NEVER reached, and the file handle stays open — this can leak
# file handles and, in write mode, may leave data unflushed/not saved to disk.
print("See comment above")
````

**Q21** Write a function `read_customers(filepath)` that reads a CSV and returns a list of dicts with `score`/`balance` cast to int, wrapping the file-open in try/except to return an empty list `[]` if the file doesn't exist. Test it on `customers.csv` (works) and `"does_not_exist.csv"` (returns empty list).

````python
# YOUR CODE HERE
````

**Solution 21**

````python
import csv

def read_customers(filepath):
    try:
        records = []
        with open(filepath, "r", newline="") as f:
            reader = csv.DictReader(f)
            for row in reader:
                records.append({
                    "name": row["name"],
                    "score": int(row["score"]),
                    "balance": int(row["balance"]),
                })
        return records
    except FileNotFoundError:
        return []

print(read_customers("customers.csv"))
print(read_customers("does_not_exist.csv"))
````

**Q22** Given `customers.csv`, filter it down to only customers with score >= 700 and write those rows to a NEW file `high_score_customers.csv` using `csv.DictWriter`, preserving the same header.

````python
# YOUR CODE HERE
````

**Solution 22**

````python
import csv

with open("customers.csv", "r", newline="") as f:
    reader = csv.DictReader(f)
    rows = [row for row in reader if int(row["score"]) >= 700]
    fieldnames = reader.fieldnames

with open("high_score_customers.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(rows)

print("Done —", len(rows), "rows written")
````

## Part E: Mini Pipeline — CSV In, Processed, CSV Out (Q23-Q25)

**Q23** Create a CSV `loans.csv` with header `["name", "score", "amount", "region"]` and 5 rows of your own realistic data using `csv.writer`. Then read it back with `DictReader` and print the number of rows loaded.

````python
# YOUR CODE HERE
````

**Solution 23**

````python
import csv

data = [
    ["name", "score", "amount", "region"],
    ["Alice", 720, 15000, "West"],
    ["Bob", 610, 8000, "East"],
    ["Cara", 800, 22000, "West"],
    ["Dave", 590, 5000, "East"],
    ["Eve", 700, 9000, "North"],
]

with open("loans.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(data)

with open("loans.csv", "r", newline="") as f:
    reader = csv.DictReader(f)
    rows = list(reader)

print(len(rows), "rows loaded")
````

**Q24** Read `loans.csv`, add a computed `"decision"` field to each row ("Approved" if score>=650 else "Declined"), and write the ENRICHED rows to `loans_decided.csv` including the new column in the header.

````python
# YOUR CODE HERE
````

**Solution 24**

````python
import csv

with open("loans.csv", "r", newline="") as f:
    reader = csv.DictReader(f)
    rows = list(reader)
    fieldnames = reader.fieldnames + ["decision"]

for row in rows:
    row["decision"] = "Approved" if int(row["score"]) >= 650 else "Declined"

with open("loans_decided.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(rows)

with open("loans_decided.csv", "r") as f:
    print(f.read())
````

**Q25** Full mini pipeline: read `loans_decided.csv`, and WITHOUT pandas, produce a summary dict `{region: {"count": N, "approved": N, "total_amount": N}}` by looping through the DictReader rows once. Print the summary.

````python
# YOUR CODE HERE
````

**Solution 25**

````python
import csv

summary = {}
with open("loans_decided.csv", "r", newline="") as f:
    reader = csv.DictReader(f)
    for row in reader:
        region = row["region"]
        if region not in summary:
            summary[region] = {"count": 0, "approved": 0, "total_amount": 0}
        summary[region]["count"] += 1
        if row["decision"] == "Approved":
            summary[region]["approved"] += 1
        summary[region]["total_amount"] += int(row["amount"])

print(summary)
````
