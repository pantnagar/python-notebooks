**Q1** Given a raw string of transactions:
```python
log = "T1:Alice:450|T2:Bob:-200|T3:Alice:120|T4:Cara:900|T5:Bob:-50"
```
Parse it and determine, per customer, whether they are net POSITIVE or net NEGATIVE overall (sum their amounts). Return a dict `{name: "Positive"/"Negative"}`.

````python
# YOUR CODE HERE
````

**Solution 1**

````python
log = "T1:Alice:450|T2:Bob:-200|T3:Alice:120|T4:Cara:900|T5:Bob:-50"

totals = {}
for entry in log.split("|"):
    _, name, amount = entry.split(":")
    totals[name] = totals.get(name, 0) + int(amount)

result = {name: ("Positive" if total >= 0 else "Negative") for name, total in totals.items()}
print(totals)
print(result)
````


**Solution 5**

````python
def count_defaults(nested):
    count = 0
    for item in nested:
        if isinstance(item, list):
            count += count_defaults(item)
        elif item is True:
            count += 1
    return count

nested = [True, False, [True, [False, True]], False, [True]]
print(count_defaults(nested))
````

**Solution 8**

**Q11** Write a function `safe_ratio(numerator, denominator)` using try/except that returns the ratio, or `float('inf')` if denominator is 0 AND numerator is positive, or `0` if BOTH are 0. Test all three cases.

````python
# YOUR CODE HERE
````

**Solution 11**

````python
def safe_ratio(numerator, denominator):
    try:
        return numerator / denominator
    except ZeroDivisionError:
        if numerator == 0:
            return 0
        return float("inf")

print(safe_ratio(10, 2))
print(safe_ratio(10, 0))
print(safe_ratio(0, 0))
````

````python
# YOUR CODE HERE
````

**Q15** Given `sequence = [100, 105, 103, 110, 108, 115, 112]` (daily balances), find the LONGEST streak of consecutive INCREASES using a single loop with running counters (no slicing tricks). Print the length of the longest streak.

````python
# YOUR CODE HERE
````

**Solution 15**

````python
sequence = [100, 105, 103, 110, 108, 115, 112]

longest = 0
current = 0
for i in range(1, len(sequence)):
    if sequence[i] > sequence[i - 1]:
        current += 1
        longest = max(longest, current)
    else:
        current = 0

print(longest)
````

**Q17** Write a function `apply_rules(record, rules)` where `rules` is a LIST OF FUNCTIONS, each taking a record dict and returning either `None` (rule doesn't apply / passes) or a STRING reason for rejection. The function should return the FIRST non-None reason found, checking rules in order, or "Approved" if none reject it. Define at least 2 rule functions and test with a record that fails the second rule.

````python
# YOUR CODE HERE
````

**Solution 17**

````python
def rule_no_defaults(record):
    if record.get("defaults", 0) > 0:
        return "Has prior defaults"
    return None

def rule_min_score(record):
    if record.get("score", 0) < 650:
        return "Score too low"
    return None

def apply_rules(record, rules):
    for rule in rules:
        result = rule(record)
        if result is not None:
            return result
    return "Approved"

rules = [rule_no_defaults, rule_min_score]
record = {"score": 600, "defaults": 0}
print(apply_rules(record, rules))
````

**Q20** Write a function `retry_with_backoff(operation, max_attempts=3)` that calls the zero-argument function `operation` up to `max_attempts` times, STOPPING as soon as it returns a truthy value (simulating a successful operation), and returning `None` if all attempts fail. Simulate `operation` as a function using a mutable list counter that only "succeeds" (returns True) on its 3rd call — use closures to track call count.

````python
# YOUR CODE HERE
````

**Solution 20**

````python
def make_flaky_operation():
    calls = [0]
    def operation():
        calls[0] += 1
        return calls[0] >= 3
    return operation

def retry_with_backoff(operation, max_attempts=3):
    for attempt in range(max_attempts):
        result = operation()
        if result:
            return result
    return None

flaky_op = make_flaky_operation()
print(retry_with_backoff(flaky_op, max_attempts=5))
````

## Full Reference Solution — Mini Project
### Don't look until you've attempted all 6 tasks yourself

````python
# Task 1: Enrich data
for a in accounts:
    a["utilization"] = round(a["balance"] / a["limit"], 3)
    if a["score"] >= 750:
        a["risk_tier"] = "Low"
    elif a["score"] >= 650:
        a["risk_tier"] = "Medium"
    else:
        a["risk_tier"] = "High"

print(accounts[0])
````

````python
# Task 2: Flag high-risk accounts
urgent_review = [a["id"] for a in accounts if a["missed_payments"] >= 2 or a["utilization"] > 0.9]
print(urgent_review)
````

````python
# Task 3: Regional summary
region_summary = {}
for a in accounts:
    r = a["region"]
    if r not in region_summary:
        region_summary[r] = {"count": 0, "total_score": 0, "total_balance": 0}
    region_summary[r]["count"] += 1
    region_summary[r]["total_score"] += a["score"]
    region_summary[r]["total_balance"] += a["balance"]

for r, info in region_summary.items():
    info["avg_score"] = round(info["total_score"] / info["count"], 1)
    del info["total_score"]

print(region_summary)
````

````python
# Task 4: Top and bottom
highest_util = sorted(accounts, key=lambda a: a["utilization"], reverse=True)[0]
lowest_score = sorted(accounts, key=lambda a: a["score"])[0]

print("Highest utilization:", highest_util["name"], highest_util["utilization"])
print("Lowest score:", lowest_score["name"], lowest_score["score"])
````

````python
# Task 5: Decision function
def decide(account):
    if account["missed_payments"] >= 3:
        return "Restrict credit"
    if account["utilization"] > 0.85:
        return "Monitor"
    if account["risk_tier"] == "Low" and account["utilization"] < 0.5:
        return "Approved for limit increase"
    return "Monitor"

decisions = {a["name"]: decide(a) for a in accounts}
print(decisions)
````

````python
# Task 6: Final report
total = len(accounts)

tier_counts = {}
for a in accounts:
    tier_counts[a["risk_tier"]] = tier_counts.get(a["risk_tier"], 0) + 1

urgent_count = len(urgent_review)

best_region = max(region_summary, key=lambda r: region_summary[r]["avg_score"])

report = f"""
CREDIT RISK SCAN REPORT
========================
Total accounts scanned: {total}
Risk tier breakdown: {tier_counts}
Accounts flagged for urgent review: {urgent_count}
Region with highest average score: {best_region} ({region_summary[best_region]['avg_score']})
"""
print(report)
````
