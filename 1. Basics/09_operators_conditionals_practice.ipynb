**Q12** Given `list_a = [1, 2, 3]` and `list_b = [1, 2, 3]` (same values, separate objects), compare them using `is` instead of `==`. Then set `list_c = list_a` and compare `list_a is list_c`. Explain the difference in a comment.

````python
# YOUR CODE HERE
list_a=[1,2,3]
list_b=[1,2,3]
list_a==list_b
list_a is list_b
````

**Solution 12**

````python
list_a = [1, 2, 3]
list_b = [1, 2, 3]
print(list_a is list_b)  # False — different objects in memory, even though values match

list_c = list_a
print(list_a is list_c)  # True — list_c points to the SAME object as list_a
# "==" checks VALUE equality, "is" checks IDENTITY (same object in memory)
````

## Part E: if / elif / else Control Flow (Q23-Q30)

**Q23** Given `score = 680`, write a full if/elif/elif/else chain assigning risk category: >=750 "Low", >=650 "Medium", >=550 "High", else "Very High". Print the result.

````python
# YOUR CODE HERE
````

**Solution 23**

**Q24** Given `score = 720` and `debt_ratio = 0.5`, write NESTED if statements (an if inside an if) to determine: if score >= 650, THEN check if debt_ratio < 0.4 for final approval, else "Review Needed"; if score < 650, "Declined" outright.

````python
# YOUR CODE HERE
````

**Solution 24**

````python
score = 720
debt_ratio = 0.5

if score >= 650:
    if debt_ratio < 0.4:
        decision = "Approved"
    else:
        decision = "Review Needed"
else:
    decision = "Declined"

print(decision)
````

**Q25** Rewrite Q24's nested logic as a FLAT set of conditions using `and`/`elif` instead of nesting, producing the same result. Compare readability in a comment.

````python
# YOUR CODE HERE
````

**Solution 25**

````python
score = 720
debt_ratio = 0.5

if score >= 650 and debt_ratio < 0.4:
    decision = "Approved"
elif score >= 650 and debt_ratio >= 0.4:
    decision = "Review Needed"
else:
    decision = "Declined"

print(decision)
# Flattening avoids deep nesting — often easier to read when there are only 2-3 conditions.
````

**Q28** Given `application_type = "joint"` and `primary_score = 700` and `co_signer_score = 600`, write conditional logic: if `application_type == "joint"`, use the AVERAGE of both scores for the decision threshold (>=650 approved); otherwise (single application) just use `primary_score`.
````python
# YOUR CODE HERE
````

**Solution 28**

````python
application_type = "joint"
primary_score = 700
co_signer_score = 600

if application_type == "joint":
    effective_score = (primary_score + co_signer_score) / 2
else:
    effective_score = primary_score

decision = "Approved" if effective_score >= 650 else "Declined"
print(effective_score, decision)
````

**Q30** Full mini pipeline: given a list of applicant dicts:
```python
applicants = [
    {"name": "Alice", "score": 720, "income": 45000, "defaults": 0},
    {"name": "Bob", "score": 610, "income": 30000, "defaults": 1},
    {"name": "Cara", "score": 800, "income": 60000, "defaults": 0},
    {"name": "Dave", "score": 660, "income": 38000, "defaults": 0},
]
```
Loop through and assign each applicant a decision using this policy: defaults > 0 → "Declined"; score >= 700 → "Approved"; score >= 650 and income >= 35000 → "Approved with conditions"; else → "Declined". Print each name with their decision.

````python
# YOUR CODE HERE
````

**Solution 30**

````python
applicants = [
    {"name": "Alice", "score": 720, "income": 45000, "defaults": 0},
    {"name": "Bob", "score": 610, "income": 30000, "defaults": 1},
    {"name": "Cara", "score": 800, "income": 60000, "defaults": 0},
    {"name": "Dave", "score": 660, "income": 38000, "defaults": 0},
]

for a in applicants:
    if a["defaults"] > 0:
        decision = "Declined"
    elif a["score"] >= 700:
        decision = "Approved"
    elif a["score"] >= 650 and a["income"] >= 35000:
        decision = "Approved with conditions"
    else:
        decision = "Declined"
    print(f"{a['name']}: {decision}")
````
