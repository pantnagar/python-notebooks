**Q1** Given `raw = "Alice:720,Bob:610,Cara:800"`, parse it into a dict `{"Alice": 720, ...}` using string splitting.

````python
# YOUR CODE HERE
````

**Solution 1**

````python
raw = "Alice:720,Bob:610,Cara:800"
pairs = raw.split(",")
result = {}
for p in pairs:
    name, score = p.split(":")
    result[name] = int(score)
print(result)
````

**Q9** Given a list of raw string records:
```python
raw_records = ["Alice,720,Approved", "Bob,610,Declined", "Cara,800,Approved"]
```
Parse each into a dict, collect them into a LIST of dicts, then use a list comprehension to extract just the names where status is "Approved".

````python
# YOUR CODE HERE
````

**Solution 9**

````python
raw_records = ["Alice,720,Approved", "Bob,610,Declined", "Cara,800,Approved"]

records = []
for row in raw_records:
    name, score, status = row.split(",")
    records.append({"name": name, "score": int(score), "status": status})

approved_names = [r["name"] for r in records if r["status"] == "Approved"]
print(records)
print(approved_names)
````

**Q12** Given `transactions = [{"customer": "Alice", "amount": 500}, {"customer": "Bob", "amount": -200}, {"customer": "Alice", "amount": 300}]`, build a dict `{customer: total}` (grouping/summing), THEN convert that dict into a sorted list of tuples `(customer, total)` ordered by total descending.

````python
# YOUR CODE HERE
````

**Solution 12**

````python
transactions = [
    {"customer": "Alice", "amount": 500},
    {"customer": "Bob", "amount": -200},
    {"customer": "Alice", "amount": 300},
]

totals = {}
for t in transactions:
    totals[t["customer"]] = totals.get(t["customer"], 0) + t["amount"]

ranked = sorted(totals.items(), key=lambda x: x[1], reverse=True)
print(totals)
print(ranked)
````

## Part C: Full Mini-Pipelines (Q17-Q25)

**Q17** Given raw applicant strings:
```python
raw = [
    "Alice|720|15000|West",
    "Bob|610|8000|East",
    "Cara|800|22000|West",
    "Dave|590|5000|East",
]
```
Build a list of dicts (parsing types correctly: score/amount as int), then compute total requested amount PER REGION as a dict.

````python
# YOUR CODE HERE
````

**Solution 17**

````python
raw = [
    "Alice|720|15000|West",
    "Bob|610|8000|East",
    "Cara|800|22000|West",
    "Dave|590|5000|East",
]

records = []
for row in raw:
    name, score, amount, region = row.split("|")
    records.append({"name": name, "score": int(score), "amount": int(amount), "region": region})

region_totals = {}
for r in records:
    region_totals[r["region"]] = region_totals.get(r["region"], 0) + r["amount"]

print(records)
print(region_totals)
````

**Q18** Using the `records` list built in Q17's logic, find the names of applicants whose score is ABOVE the average score of all applicants (a two-pass comprehension problem — you need the average first).

````python
# YOUR CODE HERE
````

**Solution 18**

````python
raw = [
    "Alice|720|15000|West",
    "Bob|610|8000|East",
    "Cara|800|22000|West",
    "Dave|590|5000|East",
]
records = [
    {"name": n, "score": int(s), "amount": int(a), "region": r}
    for n, s, a, r in (row.split("|") for row in raw)
]

avg_score = sum(r["score"] for r in records) / len(records)
above_avg = [r["name"] for r in records if r["score"] > avg_score]
print(avg_score)
print(above_avg)
````

**Q20** Given `credit_report = "score:720;utilization:35;accounts:5;delinquencies:0"`, parse this into a dict with INT values, then write a function `assess(report_dict)` that returns `"Low Risk"` if score>=700 and utilization<40 and delinquencies==0, else `"Review Needed"`. Test it.

````python
# YOUR CODE HERE
````

**Solution 20**

````python
credit_report = "score:720;utilization:35;accounts:5;delinquencies:0"

parsed = {}
for item in credit_report.split(";"):
    key, val = item.split(":")
    parsed[key] = int(val)

def assess(report_dict):
    if report_dict["score"] >= 700 and report_dict["utilization"] < 40 and report_dict["delinquencies"] == 0:
        return "Low Risk"
    return "Review Needed"

print(parsed)
print(assess(parsed))
````

**Q22** Given nested monthly data:

monthly = {

Build a dict `{name: total_across_months}` by iterating through the nested structure and summing per name (names repeat across months).

**Solution 22**

````python
monthly = {
    "Jan": {"Alice": 500, "Bob": 300},
    "Feb": {"Alice": 450, "Bob": 320},
}

totals = {}
for month, people in monthly.items():
    for name, amount in people.items():
        totals[name] = totals.get(name, 0) + amount

print(totals)
````

**Q24** Full pipeline: given raw applicant tuples `applicants = [("Alice", 720, "West"), ("Bob", 610, "East"), ("Cara", 800, "West"), ("Dave", 640, "East")]`, produce a dict where each key is a REGION and each value is a sorted LIST of names in that region with score >= 650, e.g. `{"West": ["Alice", "Cara"], "East": []}`.

````python
# YOUR CODE HERE
````

**Solution 24**

````python
applicants = [("Alice", 720, "West"), ("Bob", 610, "East"), ("Cara", 800, "West"), ("Dave", 640, "East")]

regions = {r for _, _, r in applicants}
result = {
    region: sorted([name for name, score, r in applicants if r == region and score >= 650])
    for region in regions
}
print(result)
````

**Q25** The big one — combine string parsing, dicts, sets, and comprehensions:
```python
raw_log = "C101:approved:West|C102:declined:East|C103:approved:West|C104:approved:East|C102:declined:East"
```
This is a pipe-separated log where entries can repeat (duplicates should count once). Parse it, then produce: (a) a set of unique customer IDs, (b) a dict counting APPROVED customers per region (unique customers only, no double counting repeats).

````python
# YOUR CODE HERE
````

**Solution 25**

````python
raw_log = "C101:approved:West|C102:declined:East|C103:approved:West|C104:approved:East|C102:declined:East"

entries = [tuple(e.split(":")) for e in raw_log.split("|")]
unique_entries = set(entries)  # dedupe exact repeated entries

unique_ids = {cid for cid, status, region in unique_entries}

approved_by_region = {}
for cid, status, region in unique_entries:
    if status == "approved":
        approved_by_region[region] = approved_by_region.get(region, 0) + 1

print("Unique IDs:", unique_ids)
print("Approved per region:", approved_by_region)
````
