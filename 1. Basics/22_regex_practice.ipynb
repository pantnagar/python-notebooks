# Intermediate Python — Regular Expressions (`re`)
### 25 Questions: patterns, matching, extracting, cleaning messy data

Regex looks intimidating because of the symbols, but the core idea is simple: you're describing a SHAPE of text, not exact characters. `\d{4}` means "4 digits, whatever they are" — that's it.

**This matters directly for your project:** pandas' `.str.contains()`, `.str.extract()`, and `.str.replace()` all accept regex patterns. Real credit card data has inconsistent phone formats, embedded card numbers in notes fields, malformed emails, and free-text fields you'll need to parse — regex is how.

**Cheat sheet to keep open while you work:**
- `\d` = digit, `\w` = word character (letter/digit/underscore), `\s` = whitespace
- `+` = one or more, `*` = zero or more, `?` = zero or one, `{n}` = exactly n, `{n,m}` = n to m
- `.` = any character, `^` = start of string, `$` = end of string
- `[...]` = any one of these characters, `(...)` = a capture group
- `\b` = word boundary

## Part E: Real-World Validation & Mini Pipeline (Q24-Q25)

**Q24** Write a function `is_valid_email(text)` using `re.match()` with a REASONABLE (not perfect — real email regex is famously messy) pattern: `r"^[\w.]+@[\w]+\.[a-z]+$"`. Test it against `"alice@bank.com"` (valid), `"not-an-email"` (invalid), and `"bob.smith@creditunion.org"` (valid).

````python
# YOUR CODE HERE
````

**Solution 24**

````python
import re

def is_valid_email(text):
    pattern = r"^[\w.]+@[\w]+\.[a-z]+$"
    return bool(re.match(pattern, text))

print(is_valid_email("alice@bank.com"))
print(is_valid_email("not-an-email"))
print(is_valid_email("bob.smith@creditunion.org"))
````

**Q25** The big one — a mini cleaning pipeline. Given messy raw notes from a call center log:
```python
notes = [
    "Customer Alice called, phone 555-123-4567, balance $15,000.50",
    "Bob Smith - email bob@bank.com - flagged for review",
    "Cara Lee, acct AC7890, phone: 555.987.6543, status APPROVED",
]
```
For EACH note, extract (using regex) whatever is present: any phone number (format `\d{3}[-.]?\d{3}[-.]?\d{4}` to handle both dash and dot separators), any dollar amount (`\$[\d,]+\.\d{2}`), any email, and any account code (`AC\d+`). Build a list of dicts, one per note, with keys for whatever was found (use `None` for anything not present in that note). Print the result.

````python
# YOUR CODE HERE
````

**Solution 25**

````python
import re

notes = [
    "Customer Alice called, phone 555-123-4567, balance $15,000.50",
    "Bob Smith - email bob@bank.com - flagged for review",
    "Cara Lee, acct AC7890, phone: 555.987.6543, status APPROVED",
]

results = []
for note in notes:
    phone_match = re.search(r"\d{3}[-.]?\d{3}[-.]?\d{4}", note)
    amount_match = re.search(r"\$[\d,]+\.\d{2}", note)
    email_match = re.search(r"[\w.]+@[\w]+\.[a-z]+", note)
    account_match = re.search(r"AC\d+", note)

    results.append({
        "note": note,
        "phone": phone_match.group() if phone_match else None,
        "amount": amount_match.group() if amount_match else None,
        "email": email_match.group() if email_match else None,
        "account": account_match.group() if account_match else None,
    })

for r in results:
    print(r)
````
