**Q9** Here's a classic Python gotcha. Write a BROKEN version of a function using a MUTABLE default argument:
```python
def add_flag(flag, flags=[]):

    return flags
```

**Solution 9**

**Q10** Fix the Q9 function using `flags=None` and creating a fresh list INSIDE the function body when `flags is None`. Confirm each call now starts fresh.

````python
# YOUR CODE HERE
````

**Solution 10**

````python
def add_flag(flag, flags=None):
    if flags is None:
        flags = []
    flags.append(flag)
    return flags

print(add_flag("late_payment"))
print(add_flag("high_utilization"))
````

## Part E: Recursion, Closures & Mini Challenges (Q25-Q30)

**Q25** Write a RECURSIVE function `factorial(n)` (a function that calls itself) to compute n! (used in probability calculations). Test with `factorial(5)`.

````python
# YOUR CODE HERE
````

**Solution 25**

````python
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
````

**Q26** Write a recursive function `sum_list(numbers)` that sums a list without using `sum()` — base case: empty list returns 0; recursive case: first element + sum of the rest.

````python
# YOUR CODE HERE
````

**Solution 26**

````python
def sum_list(numbers):
    if not numbers:
        return 0
    return numbers[0] + sum_list(numbers[1:])

print(sum_list([100, 200, 300, 400]))
````

**Q27** Write a CLOSURE: a function `make_risk_classifier(threshold)` that returns an INNER function which classifies a score as "Pass"/"Fail" based on the threshold captured from the outer function. Create two classifiers with different thresholds (650 and 700) and test both on the same score.

````python
# YOUR CODE HERE
````

**Solution 27**

````python
def make_risk_classifier(threshold):
    def classify(score):
        return "Pass" if score >= threshold else "Fail"
    return classify

lenient = make_risk_classifier(650)
strict = make_risk_classifier(700)

print(lenient(680))
print(strict(680))
````

**Q28** Write a function `validate_application(score, income, defaults)` that uses multiple `if` checks with EARLY RETURNS (return immediately when a rule fails, instead of nesting) to return a rejection reason string, or "OK" if all checks pass. Test with values that fail on the SECOND check to confirm it stops there.

````python
# YOUR CODE HERE
````

**Solution 28**

````python
def validate_application(score, income, defaults):
    if defaults > 0:
        return "Rejected: has defaults"
    if score < 650:
        return "Rejected: low score"
    if income < 30000:
        return "Rejected: insufficient income"
    return "OK"

print(validate_application(600, 50000, 0))
````

**Q29** Write a function `process_batch(applicants, decision_fn)` that takes a LIST of score values and a FUNCTION as arguments, applying `decision_fn` to each score and returning the list of results. Pass in a lambda as `decision_fn` when calling it. (This is exactly the pattern `.apply()` uses in pandas.)

````python
# YOUR CODE HERE
````

**Solution 29**

````python
def process_batch(applicants, decision_fn):
    return [decision_fn(a) for a in applicants]

scores = [720, 610, 800, 590]
results = process_batch(scores, lambda s: "Approved" if s >= 650 else "Declined")
print(results)
````

**Q30** Full mini pipeline using everything: write a function `evaluate(name, score, *, income=30000, **extra)` (note the `*` forcing `income` and anything after to be KEYWORD-ONLY) that:
- Returns "Declined" if score < 600
- Returns "Approved" if score >= 700 and income >= 40000
- Otherwise returns "Review"
Call it three different ways: positional name/score with default income, with income specified, and with an extra keyword arg like `region="West"` captured by `**extra`.

````python
# YOUR CODE HERE
````

**Solution 30**

````python
def evaluate(name, score, *, income=30000, **extra):
    if score < 600:
        decision = "Declined"
    elif score >= 700 and income >= 40000:
        decision = "Approved"
    else:
        decision = "Review"
    return f"{name}: {decision}"

print(evaluate("Alice", 720))
print(evaluate("Bob", 720, income=45000))
print(evaluate("Cara", 680, income=35000, region="West"))
````
