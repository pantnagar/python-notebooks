**Q26** Given a list of accounts with due dates and payment dates as strings:

```python
accounts = [
    {"name": "Alice", "due": "2024-01-05", "paid": "2024-01-04"},
    {"name": "Bob",   "due": "2024-01-05", "paid": "2024-01-20"},
    {"name": "Cara",  "due": "2024-01-05", "paid": "2024-02-15"},
]
```
Parse the dates and add a `"days_late"` field to each dict (0 if paid on time or early). Print the enriched list.

````python
# YOUR CODE HERE
````

**Solution 26**

````python
from datetime import datetime

accounts = [
    {"name": "Alice", "due": "2024-01-05", "paid": "2024-01-04"},
    {"name": "Bob",   "due": "2024-01-05", "paid": "2024-01-20"},
    {"name": "Cara",  "due": "2024-01-05", "paid": "2024-02-15"},
]

for a in accounts:
    due = datetime.strptime(a["due"], "%Y-%m-%d").date()
    paid = datetime.strptime(a["paid"], "%Y-%m-%d").date()
    days_late = max((paid - due).days, 0)
    a["days_late"] = days_late

print(accounts)
````

**Q29** Given `transactions = [{"date": "2024-01-05", "amount": 200}, {"date": "2024-01-20", "amount": 150}, {"date": "2024-02-03", "amount": 300}]`, group total amount BY MONTH (extract the "YYYY-MM" prefix as the grouping key — you don't need full date parsing for this one, string slicing works).

**Q30** Full mini pipeline: given loan records with string dates, compute a full risk view:
```python
loans = [
    {"name": "Alice", "opened": "2022-05-01", "due": "2024-01-05", "paid": "2024-01-10"},
    {"name": "Bob",   "opened": "2023-01-15", "due": "2024-01-05", "paid": "2024-02-25"},
    {"name": "Cara",  "opened": "2021-11-20", "due": "2024-01-05", "paid": "2024-01-03"},
]
```
For each loan, compute: `days_late`, and `account_age_years` (opened to today, using `(today - opened).days / 365`, rounded to 1 decimal). Print the enriched records, then print the names of anyone both `days_late > 5` AND `account_age_years < 2` (newer risky accounts).

````python
# YOUR CODE HERE
````

**Solution 30**

````python
from datetime import datetime, date

loans = [
    {"name": "Alice", "opened": "2022-05-01", "due": "2024-01-05", "paid": "2024-01-10"},
    {"name": "Bob",   "opened": "2023-01-15", "due": "2024-01-05", "paid": "2024-02-25"},
    {"name": "Cara",  "opened": "2021-11-20", "due": "2024-01-05", "paid": "2024-01-03"},
]

today = date.today()

for loan in loans:
    opened = datetime.strptime(loan["opened"], "%Y-%m-%d").date()
    due = datetime.strptime(loan["due"], "%Y-%m-%d").date()
    paid = datetime.strptime(loan["paid"], "%Y-%m-%d").date()

    loan["days_late"] = max((paid - due).days, 0)
    loan["account_age_years"] = round((today - opened).days / 365, 1)

print(loans)

risky_new_accounts = [
    l["name"] for l in loans
    if l["days_late"] > 5 and l["account_age_years"] < 2
]
print("Newer risky accounts:", risky_new_accounts)
````
