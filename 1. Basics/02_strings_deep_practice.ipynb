**Q24** Given `raw = "Name: Alice | Score: 720 | Status: Approved"`, split on `" | "` to get 3 parts, then split each part on `": "` to get clean key-value pairs. Print the result as a dictionary.

```python
# YOUR CODE HERE
```

**Solution 24**

```python
raw = "Name: Alice | Score: 720 | Status: Approved"
parts = raw.split(" | ")
result = {}
for p in parts:
    key, value = p.split(": ")
    result[key] = value
print(result)
```

**Q25** Given `messy = "  too,many,,commas,,here  "`, clean it up: strip whitespace, split on comma, and remove any empty strings from the resulting list.

```python
# YOUR CODE HERE
```

**Solution 25**

```python
messy = "  too,many,,commas,,here  "
cleaned = messy.strip().split(",")
cleaned = [x for x in cleaned if x != ""]
print(cleaned)
```
