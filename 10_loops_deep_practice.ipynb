**Q21** Given two lists `branches = ["West", "East"]` and `months = ["Jan", "Feb"]`, use nested loops to print every combination like `"West - Jan"`.

**Q22** Given `matrix = [[5, 2, 8], [1, 9, 3], [7, 4, 6]]`, use nested loops with a `running_max` variable (start at negative infinity or the first value) to find the single largest value in the entire matrix, without using `max()` directly on the whole thing.

````python
# YOUR CODE HERE
````

**Solution 22**

````python
matrix = [[5, 2, 8], [1, 9, 3], [7, 4, 6]]
running_max = matrix[0][0]
for row in matrix:
    for val in row:
        if val > running_max:
            running_max = val
print(running_max)
````

**Q23** Given `matrix = [[5, 2, 8], [1, 9, 3], [7, 4, 6]]`, use nested loops to build a NEW matrix where every value is doubled (build result row by row, appending as you go).

````python
# YOUR CODE HERE
````

**Solution 23**

````python
matrix = [[5, 2, 8], [1, 9, 3], [7, 4, 6]]
doubled = []
for row in matrix:
    new_row = []
    for val in row:
        new_row.append(val * 2)
    doubled.append(new_row)
print(doubled)
````

**Q26** Given `daily_balances = [5000, 4800, 4600, 4900, 5200]`, loop through and detect the FIRST day the balance INCREASED compared to the previous day (you'll need to track the previous value across iterations). Print which day index that was.

````python
# YOUR CODE HERE
````

**Solution 26**

````python
daily_balances = [5000, 4800, 4600, 4900, 5200]
previous = daily_balances[0]
for i in range(1, len(daily_balances)):
    current = daily_balances[i]
    if current > previous:
        print(f"First increase at day index {i}")
        break
    previous = current
````

**Q27** Simulate a loan payoff schedule with a `while` loop: `balance = 5000`, `annual_rate = 0.06`, `monthly_payment = 200`. Each month: apply interest (`balance *= (1 + annual_rate/12)`), then subtract the payment. Loop until balance <= 0, printing the balance each month, and count total months.

````python
# YOUR CODE HERE
````

**Solution 27**

````python
balance = 5000
annual_rate = 0.06
monthly_payment = 200
months = 0

while balance > 0:
    balance *= (1 + annual_rate / 12)
    balance -= monthly_payment
    months += 1
    print(f"Month {months}: {round(max(balance, 0), 2)}")

print(f"Paid off in {months} months")
````

**Q28** Given a list of customer dicts, loop through with an ACCUMULATOR pattern to build a running total AND a count simultaneously, then compute the average manually (without `sum()`):
```python
customers = [{"name": "Alice", "balance": 5000}, {"name": "Bob", "balance": 3000}, {"name": "Cara", "balance": 8000}]
```

````python
# YOUR CODE HERE
````

**Solution 28**

````python
customers = [{"name": "Alice", "balance": 5000}, {"name": "Bob", "balance": 3000}, {"name": "Cara", "balance": 8000}]

total = 0
count = 0
for c in customers:
    total += c["balance"]
    count += 1

average = total / count
print(total, count, average)
````

**Q29** Given `queue = ["C001", "C002", "C003", "C004", "C005"]` and a daily processing LIMIT of 3, use a `while` loop with `.pop(0)` to process (print) customers from the front of the queue until either the queue is empty OR the daily limit is reached — print how many are left unprocessed at the end.

````python
# YOUR CODE HERE
````

**Solution 29**

````python
queue = ["C001", "C002", "C003", "C004", "C005"]
limit = 3
processed = 0

while queue and processed < limit:
    current = queue.pop(0)
    print(f"Processing {current}")
    processed += 1

print(f"Remaining unprocessed: {len(queue)}")
````

**Q30** The big one — full loop-based mini risk scan. Given:
```python
accounts = [
    {"id": "A1", "balance": 500, "missed_payments": 0},
    {"id": "A2", "balance": -200, "missed_payments": 2},
    {"id": "A3", "balance": 1500, "missed_payments": 0},
    {"id": "A4", "balance": -50, "missed_payments": 1},
    {"id": "A5", "balance": 3000, "missed_payments": 3},
]
```
Loop through ONCE (single pass, no comprehensions) and simultaneously: (a) count how many accounts have a negative balance, (b) build a list of IDs with missed_payments >= 2, (c) track the single account with the HIGHEST balance (id + value). Print all three results at the end.

````python
# YOUR CODE HERE
````

**Solution 30**

````python
accounts = [
    {"id": "A1", "balance": 500, "missed_payments": 0},
    {"id": "A2", "balance": -200, "missed_payments": 2},
    {"id": "A3", "balance": 1500, "missed_payments": 0},
    {"id": "A4", "balance": -50, "missed_payments": 1},
    {"id": "A5", "balance": 3000, "missed_payments": 3},
]

negative_count = 0
high_risk_ids = []
top_account = accounts[0]

for a in accounts:
    if a["balance"] < 0:
        negative_count += 1
    if a["missed_payments"] >= 2:
        high_risk_ids.append(a["id"])
    if a["balance"] > top_account["balance"]:
        top_account = a

print("Negative balance accounts:", negative_count)
print("High risk (missed >= 2):", high_risk_ids)
print("Top account:", top_account["id"], top_account["balance"])
````
